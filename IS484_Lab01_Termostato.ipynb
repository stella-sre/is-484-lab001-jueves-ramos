## 5.1. Planteamiento del problema

Se quiere automatizar el control de temperatura de un ambiente mediante un sensor y un sistema de climatización (aire acondicionado y calefacción). El sistema debe decidir, en cada momento, qué acción tomar según la temperatura medida.

## 5.2. Diseño del agente (ficha PEAS)

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | La temperatura actual del ambiente, medida en °C por un sensor de temperatura. |
| **Acciones (A)** | Encender aire acondicionado, encender calefacción, no hacer nada (mantener estado). |
| **Entorno (E)** | Una habitación cerrada cuya temperatura varía por factores externos (clima, personas, equipos electrónicos) y por la propia acción del climatizador. Es parcialmente observable, dinámico y continuo. |
| **Objetivo** | Mantener la temperatura del ambiente lo más cerca posible de una temperatura deseada (temperatura_objetivo = 22°C), dentro de un margen aceptable. |
| **Medida de desempeño (P)** | Minimizar el tiempo fuera del margen aceptable; minimizar encendidos/apagados innecesarios (oscilaciones); minimizar consumo energético. |

## 5.3. De PEAS a reglas de decisión

| Condición | Acción a tomar |
|---|---|
| Si la temperatura está muy por encima de la deseada (diferencia > margen) | Encender aire acondicionado (enfriar) |
| Si la temperatura está muy por debajo de la deseada (diferencia < -margen) | Encender calefacción |
| Si la temperatura está cerca de la deseada (dentro del margen) | No hacer nada / mantener estado actual |

## 5.4. Implementación

```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Agente reactivo simple (sin memoria, sin modelo del entorno).

    PEAS:
    - Percepcion (S): temperatura_actual medida por el sensor
    - Acciones (A): "Encender A/C", "Encender calefaccion", "Mantener"
    - Entorno (E): habitacion cerrada, temperatura variable, parcialmente observable
    - Objetivo: mantener la temperatura dentro del margen respecto al objetivo
    - Medida de desempeno (P): tiempo fuera de rango, numero de cambios de estado, consumo

    Es un agente reactivo simple porque decide unicamente en base a la percepcion
    actual (temperatura_actual), sin recordar mediciones anteriores ni mantener
    ningun estado interno.
    """
    diferencia = temperatura_actual - temperatura_objetivo
    if diferencia > margen:
        return "Encender aire acondicionado"
    elif diferencia < -margen:
        return "Encender calefaccion"
    else:
        return "Mantener (sin accion)"
```

## 5.5. Simulación

```python
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 21.0, 23.0]
# Se agregaron dos valores propios:
# 21.0 -> caso limite justo en el borde del margen (diferencia = -1.0)
# 23.0 -> caso limite justo en el borde del margen (diferencia = +1.0)

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t} C -> Accion del agente: {accion}")
```

## 5.6. Reflexión

**¿Es racional este agente?**

Sí, es racional dentro de los límites de su diseño: dada la única percepción que recibe (la temperatura actual), siempre elige la acción que —según la regla fija establecida— maximiza su medida de desempeño esperada. No hay ambigüedad ni necesidad de memoria para esta tarea, porque la decisión correcta depende solo del estado actual, no del historial.

**¿Bajo qué condiciones dejaría de ser racional?**

- **Si el sensor falla o da lecturas erróneas**: el agente actuaría de forma incorrecta (por ejemplo, encendería la calefacción con el ambiente ya caliente), porque su racionalidad depende completamente de que la percepción sea fiel al entorno real.
- **Si el margen es demasiado amplio**: toleraría variaciones incómodas o ineficientes, sin cumplir realmente el objetivo aunque siga las reglas.
- **Si el margen es demasiado estrecho**: provocaría encendidos y apagados constantes (oscilación), degradando la medida de desempeño relacionada con consumo y desgaste del equipo.
- **Si el entorno cambia bruscamente** (por ejemplo, se abre una ventana): al no tener memoria ni modelo del entorno, el agente reacciona recién en el siguiente ciclo de percepción, con retraso frente a lo que un agente basado en modelos podría anticipar.
